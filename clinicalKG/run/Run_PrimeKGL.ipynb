```python
from pathlib import Path
import os
import sys

root = next(path for path in (Path.cwd(), *Path.cwd().parents)
            if (path / "clinicalKG" / "run" / "common").is_dir())
sys.path.insert(0, str(root))

DATASET = "ADNI"  # ADNI, AIBL, NACC, PPMI
SEED_COUNT = 10  # 5 or 10
TRAIN = False  # Check inputs first, then set to True
os.environ["CLINICALKG_SEED_COUNT"] = str(SEED_COUNT)
os.environ.pop("KG_ABLATION_SEEDS", None)

from clinicalKG.run.common.config import SEED_PRESETS, SEEDS
from clinicalKG.run.common.evaluation import check_inputs, train_classifiers

if SEEDS != SEED_PRESETS[SEED_COUNT]:
    raise RuntimeError("Restart the kernel after changing SEED_COUNT.")
graphs = ("PrimeKG",)
check_inputs(DATASET, graphs=graphs)
if TRAIN:
    train_classifiers(DATASET, root / "clinicalKG" / "analysis", graphs=graphs)
```
